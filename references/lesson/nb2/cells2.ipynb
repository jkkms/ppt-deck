## ① 배열 만들기

### 1번 칸 · 일주일 판매량을 배열로  ·  PPT 6쪽

```python
import numpy as np

# 월요일부터 일요일까지
sales = np.array([12, 15, 9, 20, 18, 25, 30])
print(sales * 2)   # 칸마다 두 배
```

### 2번 칸 · 평균과 가장 많이 판 날  ·  PPT 7쪽

```python
print(sales.mean().round(1))
print(sales.argmax())   # 0 = 월요일
```

## ② 그림으로 보기

### 3번 칸 · 꺾은선으로 그려 보기

```python
days = ['월', '화', '수', '목', '금', '토', '일']
plt.plot(days, sales, marker='o')
plt.ylabel('판매량 (개)')
plt.show()
```
